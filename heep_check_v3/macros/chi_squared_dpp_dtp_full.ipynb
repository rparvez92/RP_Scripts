```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import colormaps as cm
from scipy.optimize import minimize

plt.rcParams['xtick.labelsize'] = 16
plt.rcParams['ytick.labelsize'] = 16

heep_slope = np.array([
    [-24.63, -10.73,  0.00,  0.00],
    [  0.00,  -6.71,  0.00,  -4.73],
    [  -4.73,  4.75,  0.00,  4.81],
    [ 4.75,  4.73,  -4.81,  0.00]
])
```

## 6.720 GeV

```python
diffs = pd.read_csv('∆s_6p720_full.csv')


p0 = (0,0,0,0) # initial guess for all offsets


def chi_squared(p):
    dθe, dpe, dθp, dpp = p[0], p[1], p[2], p[3]

    chi2 = 0

    d_offsets = np.array([dθe, dpe, dθp, dpp])
    Δcalc = heep_slope @ d_offsets

    row_W  = diffs[diffs['variable']=='W'].iloc[0]
    row_Em = diffs[diffs['variable']=='Em'].iloc[0]
    row_pz = diffs[diffs['variable']=='pz'].iloc[0]
    row_py = diffs[diffs['variable']=='py'].iloc[0]

    Δdata = np.array([
        row_W['mu'], row_Em['mu'], row_pz['mu'], row_py['mu']
    ])

    sigmas = np.array([
        row_W['mu_err'], row_Em['mu_err'],
        row_pz['mu_err'], row_py['mu_err']
    ])

    chi2 += np.sum(((Δdata - Δcalc) / sigmas)**2)

    return chi2
```

```python
result = minimize(
    chi_squared,
    p0,
    method='Nelder-Mead',
)

best_params = result.x
min_chi2 = chi_squared(best_params)
```

```python
print(result.success)
print(result.message)
print(result.nit)   # number of iterations
print(result.nfev)  # function evaluations
```

```python
print("Optimal offsets:")

print(f"dθe: {best_params[0]:.4f} %")
print(f"dpe: {best_params[1]:.4f} mrad")
print(f"dθp: {best_params[2]:.4f} mrad")
print(f"dpp: {best_params[3]:.4f} %")
print(f"Minimum chi-squared: {min_chi2:.4f}")
```

```python
diffs_6p720 = pd.read_csv('∆s_6p720_full.csv')
diffs_6p6048 = pd.read_csv('∆s_6p6048_full.csv')
diffs_6p384 = pd.read_csv('∆s_6p384_full.csv')

diffs_total = pd.concat([diffs_6p720, diffs_6p6048, diffs_6p384], ignore_index=True)

diffs_total['residual'] = diffs_total['mu'] - diffs_total['heepcheck']

fig, ax = plt.subplots(figsize=(19, 9))

ax.errorbar(diffs_total['delta_center'][(diffs_total['variable'] == 'W')], diffs_total['residual'][(diffs_total['variable'] == 'W')], yerr=diffs_total['mu_err'][(diffs_total['variable'] == 'W')], fmt='o', markersize=7, color=cm.get_cmap('Dark2')(0), label='W')
ax.errorbar(diffs_total['delta_center'][(diffs_total['variable'] == 'Em')], diffs_total['residual'][(diffs_total['variable'] == 'Em')], yerr=diffs_total['mu_err'][(diffs_total['variable'] == 'Em')], fmt='^', markersize=7, color=cm.get_cmap('Dark2')(1), label='Em')
ax.errorbar(diffs_total['delta_center'][(diffs_total['variable'] == 'pz')], diffs_total['residual'][(diffs_total['variable'] == 'pz')], yerr=diffs_total['mu_err'][(diffs_total['variable'] == 'pz')], fmt='v', markersize=7, color=cm.get_cmap('Dark2')(2), label='pz')
ax.errorbar(diffs_total['delta_center'][(diffs_total['variable'] == 'py')], diffs_total['residual'][(diffs_total['variable'] == 'py')], yerr=diffs_total['mu_err'][(diffs_total['variable'] == 'py')], fmt='s', markersize=7, color=cm.get_cmap('Dark2')(3), label='py')
ax.set_title(r'Measured - heepcheck offsets ($\chi^2$ minimization)', fontsize=15, fontweight='bold')
ax.set_xlabel('δ bin center', fontsize=14, fontweight='bold')
ax.set_ylabel('Residual (MeV)', fontsize=14, fontweight='bold')
ax.legend(fontsize=14, loc='upper right')

ax.fill_betweenx([-5, 5], -0.5, 2.5, color='blue', alpha=0.2)
ax.fill_betweenx([-5, 5], 1.5, 4.5, color='red', alpha=0.2)
ax.fill_betweenx([-5, 5], 4.5, 7.5, color='green', alpha=0.2)


ax.grid(True, which='both', linestyle='--', linewidth=0.5)

for spine in ax.spines.values():
        spine.set_linewidth(3)
        spine.set_color('black')
```

```python
delta_centers = [1,3,6]

def chi_squared(p):
    dpe_list = p[0:3]
    dθe_list = p[3:6]
    dθp, dpp = p[6:8]

    chi2 = 0

    for i, dc in enumerate(delta_centers):

        dθe = dθe_list[i]
        dpe = dpe_list[i]

        d_offsets = np.array([dθe, dpe, dθp, dpp])
        Δcalc = heep_slope @ d_offsets

        row_W  = diffs_total[(diffs_total['variable']=='W')  & (diffs_total['delta_center']==dc)].iloc[0]
        row_Em = diffs_total[(diffs_total['variable']=='Em') & (diffs_total['delta_center']==dc)].iloc[0]
        row_pz = diffs_total[(diffs_total['variable']=='pz') & (diffs_total['delta_center']==dc)].iloc[0]
        row_py = diffs_total[(diffs_total['variable']=='py') & (diffs_total['delta_center']==dc)].iloc[0]

        Δdata = np.array([
            row_W['mu'], row_Em['mu'], row_pz['mu'], row_py['mu']
        ])

        sigmas = np.array([
            row_W['mu_err'], row_Em['mu_err'],
            row_pz['mu_err'], row_py['mu_err']
        ])

        chi2 += np.sum(((Δdata - Δcalc) / sigmas)**2)

    return chi2
```

```python
# p0 = (0,0,0,0,0,0,0,0) # initial guess for all offsets

p0 = (0.9869, 1.0048, 1.4619,         # initial  dpe offsets for all 6 delta centers
      -0.8619, -0.6471, -1.1178,      # initial dθe offsets for all 6 delta centers
    #   -1.8051, 1.1197)                # initial dθp and dpp offsets
    0, 0) 
result = minimize(
    chi_squared,
    p0,
    method='Nelder-Mead',
)

best_params = result.x
min_chi2 = chi_squared(best_params)
# 
```

```python
print(result.success)
print(result.message)
print(result.nit)   # number of iterations
print(result.nfev)  # function evaluations
```

```python
print("Optimal offsets:")
for i in range(3):  
    print(f"dθe{i+1}: {best_params[i+3]:.4f} mrad")

for i in range(3):  
    print(f"dpe{i+1}: {best_params[i]:.4f} %")

print(f"dθp: {best_params[6]:.4f} mrad")
print(f"dpp: {best_params[7]:.4f} %")
print(f"Minimum chi-squared: {min_chi2:.4f}")
```
